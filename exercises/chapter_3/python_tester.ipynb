# Testing problem 3.3

```python
filename = "xy.dat"

# FOrmat is strings in columns of x,y
data = np.loadtxt(filename, dtype=str)
x = []
y = []
for point in data[:len(data)-1]:
    # x is everything before the comma, y is everything after the comma
    x.append(float(point.split(",")[0]))
    y.append(float(point.split(",")[1]))

x = np.linspace(0, 1, 100)
y = np.exp(-x)
plt.plot(x, y, label = "test")
plt.scatter(x, y, label = "data")
plt.legend()
```
